```python
dates=pd.read_csv('../data/clean_data/dim_date_updated_cleaned.csv')
```

```python
df=transactions.merge(
    dates,
    on='date_id',
    how="left"
)


df=df.merge(
    worker,
    on='worker_id',
    how="left"
)


df=df.merge(
    channel,
    on='channel_id',
    how="left"
)

df=df.merge(
    market,
    on='market_id',
    how="left"
)

```
